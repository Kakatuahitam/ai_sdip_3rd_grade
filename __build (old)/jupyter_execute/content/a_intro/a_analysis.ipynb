## Learning Goals Analysis (_Analisis Capaian Pembelajaran_)

Before going to know Artificial Intelligence Learning Goals Analysis, let us
understand first how Artificial Intelligence characterize itself. Artificial
Intelligence subject has been introduced for children with Erasmus+ project
named "Artificial Intelligence Education for Children" that is supported by
European Commission for Erasmus+ Program.

Its objective are to create education system that is expected to fulfill the
coding related job vacancies and to prepare Artificial Intelligence readiness
in three aspects; (1) Artificial Intelligence is being put at the service of
European Citizens so it should be more encouraged to be researched, (2)
Artificial Intelligence will be key factor to economic development, and (3)
pursuing competitive Artificial Intelligence for countries or region. So, the
project is intended to develop educational materials aimed to children for the
teachers who teaches Artificial Intelligence fundamentals based on visual coding
experiences with various games and activities.

## Artificial Intelligence Learning Goals

Based on reference from Artificial Intelligence book from Turkey and kind of
kids nature in B and C Phase, kids are being so active in their motoric so it
gives option to enhance everyday learning in moving bodies while percepting
the subject we learn on the class.

To keep the learning outline always in a good way of easy and self learning,
this Artificial Intelligence subject has these main Element:

1. Human and Computer
1. Robotics and Artificial Intelligence
1. Artificial Intelligence Ethic

Furthermore, this Artificial Intelligence for Phase B has these Learning
Goals:

### Artificial Intelligence Learning Goals for Phase B

1. To know what is Computer and its other form
1. To know how computer works
1. To know how computer interact with human
1. To know what is Artificial Intelligence really is
1. ...

### Artificial Intelligence Learning Goals for Phase C

1. To know and create computer program for helping daily activities
1. To understand how Artificial Intelligence is developed and utilized in
programs
1. To understand how to build robotic tools and utilize it for daily activities
1. To understand how to integrate Artificial Intelligence on Robotic Tools
implementation
