
See [](#my-math-label) for an equation!

Learning Outcome Analysis (Analisis Capaian Pembelajaran) is the main goal of
that SD Islam Pembangunan aims to be reached for the children to realize one of
Graduate Student Profile that is [_Lulusan yang Cakap Teknologi Digital_](https://sites.google.com/pembangunan.sch.id/papan-informasi-sdip/visi-misi-profile-lulusan).

The points of the profile are:
- _Menguasai dasar-dasar algoritma_
- _Menguasai dasar-dasar ilmu komputer_
- _Menguasai dasar-dasar sains_
- _Menguasai dasar-dasar program_
- _Menguasai aplikasi teknologi sederhana_
- _Terbiasa memanfaatkan teknologi digital_

Those points stated in Graduate Student Profile is the base of Artificial
Intelligence Learning Outcome. Strengthen by statement that states this era is
needing Artificial Intelligence agent that will fill the gap in the knowledge
base of Artificial Intelligence in the public and professional space. So that
learning Artificial Intelligence is enforced since early school stage in this
case elementary school.

Students will be taught Artificial Intelligence subject as local content that
SD Islam Pembangunan pursue to educate and enlight elementary students to know,
learn, adapt, utilize, and even create Artificial Intelligence products since
B and C Phase as a part of implementing Kurikulum Merdeka.

To create more enjoy and simple learning for B (3rd and 4th Grade) and C (5th
and 6th Grade) Phase, the learning session will be conducted in two main form.
Those are plugged and Unplugged sessions that will be changes transiently from
70% of unplugged session and 30% plugged session starts from 3rd Grade to the
70% of plugged session and 30% unplugged session. It is designed to facilitate
students to fulfill their body movement needs as the children is designed to
keep playing with each others actively in form of physical activities. And it is
also helps to regulate screen time management for students that is limited to
prevent overtime in screen activities. Finally, the academic output of Learning
will be improved as their physical activity has been enhanced cognitive
function such as attention, memory, and problem solving. {cite}`hiroshi_2024`.

---
```{bibliography}
:style: unsrtalpha
```
